# pandas: небольшой аналитический результат


Показать типы, группировку, объединение, сводную таблицу и временной ряд в pandas.

Нужен пакет `pandas`; все данные создаются в памяти и не требуют внешних сервисов.

Ограничения: Когда результат уже мал и требуется локальная проверка или подготовка отчёта. Не используйте `toPandas()` для полной таблицы и проверяйте типы дат перед группировкой.


## Основные понятия
pandas подходит для ограниченного результата. Перед переносом данных сначала уменьшайте объём на стороне Spark или Trino.


```python
import pandas as pd
```

## Примеры


**1. Типы и пропуски**

Проверьте типы и число пропусков в малом кадре.



```python
frame = pd.DataFrame({
    "match_id": [1, 2, 3], 
    "duration": [1800, None, 2400],
    "winner": ["Radiant", "Dire", "Radiant"]
})

print(frame.dtypes)
print(frame.isna().sum())
```

**2. Группировка**

Сводите длительность по категории победителя.



```python
frame = pd.DataFrame({
    "winner": ["Radiant", "Dire", "Radiant"], 
    "duration": [1800, 2400, 2100]
})

print(frame.groupby("winner", as_index=False).agg(matches=("winner", "size"), average_duration=("duration", "mean")))
```

**3. Объединение**

Добавьте справочное название к фактам.



```python
facts = pd.DataFrame({"hero_id": [1, 2, 1], "kills": [4, 7, 2]})
heroes = pd.DataFrame({"hero_id": [1, 2], "name": ["A", "B"]})

print(facts.merge(heroes, on="hero_id", how="left", validate="many_to_one"))
```

**4. Сводная таблица**

Преобразуйте малую таблицу в формат для отчёта.



```python
frame = pd.DataFrame({
    "winner": ["Radiant", "Dire", "Radiant"], 
    "season": [1, 1, 2], 
    "matches": [10, 8, 12]
})

print(frame.pivot_table(index="season", columns="winner", values="matches", fill_value=0))
```

**5. Временной ряд**

Покажите изменение числа событий по дням.



```python
frame = pd.DataFrame({
    "day": pd.to_datetime(["2024-01-01", "2024-01-01", "2024-01-02"]), 
    "events": [3, 5, 4]
})

print(frame.set_index("day").resample("D")["events"].sum().rolling(2, min_periods=1).mean())
```
