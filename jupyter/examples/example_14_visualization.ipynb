# Визуализация: малый агрегированный результат


Построить столбцы, распределение и связь показателей с читаемыми подписями.

Нужны `pandas` и `matplotlib`; малые наборы создаются в памяти, внешние сервисы не нужны.

Ограничения: Для исследования и иллюстрации уже агрегированных данных. График не исправляет неверную агрегацию; избегайте обрезанных осей и слишком большого числа категорий.


## Основные понятия
Spark агрегирует, pandas хранит малый результат, matplotlib рисует. Ось, единицы и уровень детализации должны быть явными.


```python
import matplotlib.pyplot as plt
import pandas as pd
```

## Примеры


**1. Столбцы**

Покажите среднюю длительность по победившей стороне.



```python
frame = pd.DataFrame({
    "winner": ["Radiant", "Dire"], 
    "minutes": [32.1, 30.4]
})
axis = frame.plot.bar(
    x="winner", y="minutes", legend=False, 
    color=["#4f8f70", "#c65d4b"], 
    figsize=(7, 4)
)
axis.set_ylabel("Минуты")
axis.set_title("Средняя длительность")
plt.tight_layout()
plt.show()
```

**2. Распределение**

Посмотрите форму распределения длительности на малом наборе.



```python
frame = pd.DataFrame({"minutes": [20, 25, 28, 30, 31, 35, 42]})
frame.plot.hist(y="minutes", bins=4, legend=False, figsize=(7, 4))
plt.xlabel("Минуты")
plt.tight_layout()
plt.show()
```

**3. Точечная диаграмма**

Сравните длительность и общий счёт с прозрачными подписями осей.



```python
import matplotlib.pyplot as plt
import pandas as pd

frame = pd.DataFrame({"minutes": [20, 25, 30, 35], "total_kills": [20, 35, 28, 44]})
frame.plot.scatter(x="minutes", y="total_kills", figsize=(7, 4), alpha=0.8)
plt.xlabel("Длительность, минуты")
plt.ylabel("Общий счёт")
plt.tight_layout()
plt.show()
```

**4. Подписи и единицы**

Проверьте, что таблица для графика содержит короткие названия и единицы измерения.



```python
frame = pd.DataFrame({"category": ["A", "B"], "average_minutes": [10.5, 12.0]})

print(frame.rename(columns={"average_minutes": "Среднее, минуты"}).to_string(index=False))
```

**5. Проверка малой выдачи**

Зафиксируйте число строк, которое переносится в график.



```python
frame = pd.DataFrame({"category": list("ABCDE"), "value": [1, 4, 2, 5, 3]})
assert len(frame) <= 20

print("Строк для графика:", len(frame))
```
