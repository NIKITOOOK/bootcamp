# Spark: контекст и каталог


Научиться открывать краткоживущий контекст Spark и исследовать доступные объекты.

Ожидаются каталоги `lakehouse` и пространства `raw`, `stage`, `mart`. Все ячейки создают собственный контекст.

Ограничения: Когда нужно быстро проверить подключение, каталог или схему перед исследованием. Не переносите DataFrame между ячейками с разными контекстами и не вызывайте `collect()` без ограничения.


## Основные понятия
Контекст управляет приложением и закрывается после ячейки. Каталог содержит пространства имён, таблицы и их схемы. Действие запускает вычисление.


## Примеры


**1. Свойства приложения**

Проверьте версию, идентификатор приложения и настройки без чтения таблиц.

Сопоставьте поля, значения и размер выдачи с ожидаемым поведением операции; при расхождении сначала проверьте схему.


```python
from lakehouse_spark import SparkSessionContext
from pyspark.sql import functions as F


with SparkSessionContext('example-context-properties') as spark:
    print(spark.version)
    print(spark.sparkContext.applicationId)
    print(spark.sparkContext.master)
    print(spark.conf.get("spark.sql.shuffle.partitions"))
```

**2. Пространства имён**

Список пространств помогает понять, какие области доступны текущему каталогу.

Сопоставьте поля, значения и размер выдачи с ожидаемым поведением операции; при расхождении сначала проверьте схему.


```python
from lakehouse_spark import SparkSessionContext
from pyspark.sql import functions as F


with SparkSessionContext('example-context-namespaces') as spark:
    spark.sql("show namespaces in lakehouse").show(30, truncate=False)
```

**3. Таблицы пространства**

Ограниченный список таблиц показывает имена без чтения их содержимого.

Сопоставьте поля, значения и размер выдачи с ожидаемым поведением операции; при расхождении сначала проверьте схему.


```python
from lakehouse_spark import SparkSessionContext
from pyspark.sql import functions as F


with SparkSessionContext('example-context-tables') as spark:
    spark.sql("show tables in lakehouse.raw").show(50, truncate=False)
```

**4. Схема таблицы**

Расширенное описание полезно перед составлением запроса и проверкой типов.

Сопоставьте поля, значения и размер выдачи с ожидаемым поведением операции; при расхождении сначала проверьте схему.


```python
from lakehouse_spark import SparkSessionContext
from pyspark.sql import functions as F


with SparkSessionContext('example-context-schema') as spark:
    spark.sql("describe table extended lakehouse.raw.pg__matches").show(100, truncate=False)
```

**5. Небольшой просмотр**

Прочитайте несколько строк после явного ограничения количества строк.

Сопоставьте поля, значения и размер выдачи с ожидаемым поведением операции; при расхождении сначала проверьте схему.
